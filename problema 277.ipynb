```python
def encontrar_maximo_manual(lista_numeros):
    """
    Devuelve el valor máximo de una lista recorriendo los elementos manualmente.
    """
    if not lista_numeros:
        return None  # Devuelve None si la lista está vacía

    # Inicializar el máximo con el primer elemento de la lista
    maximo_actual = lista_numeros[0]

    # Iterar sobre el resto de los elementos de la lista
    for numero in lista_numeros[1:]:
        # Si el número actual es mayor que el máximo actual, actualizar el máximo
        if numero > maximo_actual:
            maximo_actual = numero
            
    return maximo_actual

# --- Ejemplo de Uso ---
numeros = [73, 11, 55, 99, 14, 88]
maximo = encontrar_maximo_manual(numeros)

print(f"Lista: {numeros}")
print(f"El valor máximo (implementación manual) es: {maximo}")
```

Bucle for: El bucle itera sobre una sub-lista creada con el slicing lista_numeros[1:].
 Esto significa que el bucle comienza desde el segundo elemento de la lista (índice 1) y recorre hasta el final. Se excluye el primer elemento porque ya lo usamos para inicializar maximo_actual.

Comparación: Dentro del bucle, en cada iteración:

Se compara el valor del elemento actual (numero) con el maximo_actual registrado.

Si numero es mayor (if numero > maximo_actual), la variable maximo_actual se actualiza con el valor de numero.
Retorno: Una vez que el bucle termina (después de haber comparado todos los elementos), maximo_actual contiene el valor más grande de toda la lista, y este valor es devuelto.
