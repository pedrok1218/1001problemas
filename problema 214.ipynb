```python
import random

def juego_adivina_numero():
    """
    Genera un número secreto entre 1 y 100 y permite al usuario adivinarlo,
    dándole pistas hasta que acierte.
    """
    # 1. Configuración del juego
    numero_secreto = random.randint(1, 100)
    intentos = 0
    adivinado = False

    print("--- ¡Bienvenido al juego Adivina el Número! ---")
    print("He elegido un número entre 1 y 100. ¡Adivínalo!")

    # 2. Bucle principal de juego
    while not adivinado:
        try:
            # Solicitar la entrada al usuario
            entrada = input("Intenta adivinar el número: ")
            intento_usuario = int(entrada)
            intentos += 1

            # 3. Comprobar límites
            if intento_usuario < 1 or intento_usuario > 100:
                print("⚠️ ¡ERROR! El número debe estar entre 1 y 100. Intenta de nuevo.")
                continue # Vuelve al inicio del bucle sin contar el intento como válido

            # 4. Dar pistas
            if intento_usuario < numero_secreto:
                print(f"{intento_usuario} → Es **mayor**")
            elif intento_usuario > numero_secreto:
                print(f"{intento_usuario} → Es **menor**")
            else:
                # 5. ¡Acierto!
                adivinado = True
                print(f"\n{intento_usuario} → ¡**Correcto**! 🎉")

        except ValueError:
            print("❌ Entrada inválida. Por favor, introduce un número entero.")
        
    # 6. Fin del juego
    print(f"\n¡Felicidades! Adivinaste el número {numero_secreto} en {intentos} intentos.")

# Iniciar el juego
juego_adivina_numero()
```

Importación: Se utiliza import random para acceder a la función random.randint(1, 100), que elige un entero aleatorio dentro del rango inclusivo.

Bucle while: La condición principal es while not adivinado. Mientras la variable booleana adivinado sea False, el programa seguirá pidiendo un número al usuario.

Manejo de Errores: Se usa un bloque try...except ValueError para manejar casos en los que el usuario introduce texto en lugar de un número, evitando que el programa colapse.
