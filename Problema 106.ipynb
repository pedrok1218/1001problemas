```python
# Solicita la primera cadena de texto
cadena1 = input("Introduce la primera cadena de texto: ")

# Solicita la segunda cadena de texto
cadena2 = input("Introduce la segunda cadena de texto: ")

# Concatena las dos cadenas usando el operador +
# Se añade un espacio entre ellas para una mejor legibilidad, 
# pero la concatenación pura sería solo cadena1 + cadena2
resultado = cadena1 + " " + cadena2 

# Muestra el resultado
print("\nLa cadena concatenada es:")
print(resultado)
```

cadena1 = input(...): La función input() pide al usuario que introduzca texto. Lo que el usuario escribe se almacena en la variable cadena1 como una cadena de texto (str).

cadena2 = input(...): Hace lo mismo para la segunda cadena.

resultado = cadena1 + " " + cadena2: Aquí es donde ocurre la concatenación.

El operador + se utiliza para unir dos o más cadenas de texto.

En este caso, se une cadena1, un espacio " " (para separar las palabras), y cadena2.

print(resultado): Imprime la cadena de texto final resultante.
