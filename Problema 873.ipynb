```python
class NodoLibro:
    """
    Representa un nodo en la lista enlazada (un libro).
    """
    def __init__(self, titulo, autor=None):
        self.titulo = titulo
        self.autor = autor
        self.siguiente = None  # Referencia al siguiente nodo

    def __str__(self):
        """Devuelve la representación del libro como cadena."""
        if self.autor:
            return f'"{self.titulo}" de {self.autor}'
        return f'"{self.titulo}"'

class Biblioteca:
    """
    Gestiona la lista enlazada de libros.
    """
    def __init__(self):
        self.cabeza = None  # El inicio de la lista enlazada

    def agregar_libro(self, titulo, autor=None):
        """Agrega un libro al inicio de la lista."""
        nuevo_nodo = NodoLibro(titulo, autor)
        # El nuevo nodo apunta a la cabeza actual
        nuevo_nodo.siguiente = self.cabeza
        # El nuevo nodo se convierte en la nueva cabeza
        self.cabeza = nuevo_nodo
        print(f'Libro "{titulo}" agregado exitosamente.')

    def eliminar_libro(self, titulo):
        """Elimina la primera ocurrencia de un libro por título."""
        actual = self.cabeza
        previo = None

        # Caso 1: El libro a eliminar es la cabeza
        if actual and actual.titulo == titulo:
            self.cabeza = actual.siguiente
            print(f'El libro "{titulo}" fue eliminado de la biblioteca.')
            return

        # Caso 2: Buscar el libro en el resto de la lista
        while actual and actual.titulo != titulo:
            previo = actual
            actual = actual.siguiente

        # Si el libro no fue encontrado
        if actual is None:
            print(f'Error: El libro "{titulo}" no se encontró para eliminar.')
            return

        # El libro fue encontrado: Enlazar el nodo previo con el siguiente al actual
        if previo:
            previo.siguiente = actual.siguiente
        
        print(f'El libro "{titulo}" fue eliminado de la biblioteca.')

    def buscar_libro(self, titulo):
        """Busca un libro por título e informa si fue encontrado."""
        actual = self.cabeza
        
        while actual:
            if actual.titulo == titulo:
                print(f'El libro "{titulo}" fue encontrado en la biblioteca.')
                return True
            actual = actual.siguiente
        
        print(f'El libro "{titulo}" no fue encontrado en la biblioteca.')
        return False

    def mostrar_libros(self):
        """Muestra todos los libros en la biblioteca."""
        print('Lista de libros en la biblioteca:')
        actual = self.cabeza
        if actual is None:
            print('La biblioteca está vacía.')
            return

        while actual:
            print(f'{actual}')
            actual = actual.siguiente

# ----------------------------------------------------------------------
## Programa Principal y Datos de Prueba
# ----------------------------------------------------------------------

# 1. Inicializar la biblioteca
mi_biblioteca = Biblioteca()

# 2. Agregar libros
mi_biblioteca.agregar_libro("Cien años de soledad", "Gabriel García Márquez")
mi_biblioteca.agregar_libro("La Odisea", "Homero")

# 3. Buscar libro
mi_biblioteca.buscar_libro("Cien años de soledad")

# 4. Eliminar libro
mi_biblioteca.eliminar_libro("La Odisea")

# 5. Mostrar lista completa
mi_biblioteca.mostrar_libros()

# ----------------------------------------------------------------------
```
