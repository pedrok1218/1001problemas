```python
def encontrar_mayor_menor():
    """
    Lee una secuencia de números del usuario y encuentra el mayor y el menor.
    """
    try:
        # 1. Obtener la cantidad de números a ingresar
        cantidad = int(input("¿Cuántos números vas a ingresar en la secuencia? "))
        
        if cantidad <= 0:
            print("❌ Por favor, ingresa una cantidad de números válida (mayor que 0).")
            return

        # Inicializar variables para el mayor y el menor
        # Usamos 'None' para asegurar que el primer número leído se convierta
        # automáticamente en el mayor y el menor iniciales.
        numero_mayor = None
        numero_menor = None

        # 2. Iniciar el ciclo FOR para leer los números
        # El ciclo se repite 'cantidad' veces (de 0 a cantidad-1)
        for i in range(cantidad):
            
            # 3. Leer el número actual
            entrada = input(f"Ingresa el número {i + 1}: ")
            numero_actual = float(entrada) # Convertir a float para permitir decimales

            # 4. Actualizar el Mayor y el Menor
            
            # Caso de la primera iteración (inicialización):
            if numero_mayor is None:
                numero_mayor = numero_actual
                numero_menor = numero_actual
            
            # Caso de las siguientes iteraciones:
            else:
                # Verificar el mayor
                if numero_actual > numero_mayor:
                    numero_mayor = numero_actual
                
                # Verificar el menor
                if numero_actual < numero_menor:
                    numero_menor = numero_actual

        # 5. Mostrar los resultados
        print("\n--- Resultados ---")
        print(f"✅ El número mayor ingresado es: {numero_mayor}")
        print(f"✅ El número menor ingresado es: {numero_menor}")

    except ValueError:
        print("❌ Error: Por favor, ingresa solo números enteros o decimales válidos.")
    except Exception as e:
        print(f"Ocurrió un error: {e}")

# Llamar a la función principal para ejecutar el programa
encontrar_mayor_menor()
```

Inicialización: Las variables numero_mayor y numero_menor se inicializan a None. Esta técnica es vital: asegura que el primer número que el usuario ingrese se establezca inmediatamente como el valor inicial para ambos, mayor y menor, independientemente de cuán grande o pequeño sea.

Ciclo for: Se utiliza range(cantidad) para garantizar que el programa se ejecute exactamente la cantidad de veces que el usuario especificó. Esto cumple con el requisito de usar un ciclo for.

Conversión: La entrada del usuario se convierte a float (numero_actual = float(entrada)) para permitir que el programa maneje tanto números enteros como decimales.
