```python
def es_primo(numero):
    """
    Verifica si un número entero es primo.
    Un número es primo si es mayor que 1 y solo es divisible por 1 y por sí mismo.
    """
    if numero <= 1:
        return False
    # Optimizamos la verificación: solo necesitamos verificar divisores hasta la 
    # raíz cuadrada del número.
    for i in range(2, int(numero**0.5) + 1):
        if numero % i == 0:
            return False  # Se encontró un divisor, no es primo
    return True # Si no se encontraron divisores, es primo

def listar_primos_en_rango():
    """
    Solicita un rango al usuario y lista todos los números primos dentro de ese rango.
    """
    print("--- Buscador de Números Primos en un Rango ---")
    
    try:
        # 1. Solicitar el rango al usuario
        inicio = int(input("Ingresa el número inicial del rango (e.g., 1): "))
        fin = int(input("Ingresa el número final del rango (e.g., 100): "))
        
        # 2. Validar que el inicio sea menor o igual al fin
        if inicio > fin:
            print("❌ El número inicial no puede ser mayor que el final. Intercambiando valores.")
            inicio, fin = fin, inicio # Intercambia los valores
            
        print(f"\nBuscando números primos entre {inicio} y {fin}...")
            
        primos_encontrados = []
        
        # 3. Recorrer el rango e identificar primos
        # Se usa fin + 1 porque range() excluye el límite superior.
        for numero in range(inicio, fin + 1):
            if es_primo(numero):
                primos_encontrados.append(numero)
                
        # 4. Mostrar el resultado
        if primos_encontrados:
            print(f"\n✅ Números primos encontrados ({len(primos_encontrados)} en total):")
            print(primos_encontrados)
        else:
            print("\n😔 No se encontraron números primos en el rango especificado.")
            
    except ValueError:
        print("❌ Error: Por favor, ingresa solo números enteros válidos.")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")

# Ejecutar el programa
listar_primos_en_rango()
```

Función listar_primos_en_rango():

Pide el inicio y el fin del rango.

Utiliza un bucle for (for numero in range(inicio, fin + 1)) para iterar sobre todos los enteros en el rango definido.

Llama a la función es_primo() para cada número.

Si es_primo() devuelve True, añade el número a la lista primos_encontrados.

Finalmente, imprime todos los resultados.
