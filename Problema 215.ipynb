```python
def obtener_nombre_del_mes():
    """
    Solicita un número de mes del 1 al 12 e imprime el nombre del mes.
    """
    
    # Lista de nombres de meses (el índice 0 se deja vacío o se usa para un placeholder, 
    # ya que los meses van del 1 al 12)
    meses = [
        "Inválido",  # Índice 0 (Placeholder)
        "enero",
        "febrero",
        "marzo",
        "abril",
        "mayo",
        "junio",
        "julio",
        "agosto",
        "septiembre",
        "octubre",
        "noviembre",
        "diciembre"
    ]

    while True:
        try:
            # 1. Solicitar la entrada del usuario
            entrada = input("Ingrese un número de mes (1-12): ")
            
            # 2. Convertir la entrada a entero
            numero_mes = int(entrada)

            # 3. Validar el rango
            if 1 <= numero_mes <= 12:
                # 4. Imprimir el resultado
                nombre_mes = meses[numero_mes]
                print(f"Mes: {nombre_mes}")
                break  # Salir del bucle si la entrada es válida
            else:
                print("Error: El número debe estar entre 1 y 12.")

        except ValueError:
            # Manejo de error si la entrada no es un número entero
            print("Error: Por favor, ingrese un valor numérico entero.")
        except Exception as e:
            # Manejo de cualquier otro error inesperado
            print(f"Ocurrió un error inesperado: {e}")

# Ejecutar la función
obtener_nombre_del_mes()
```
