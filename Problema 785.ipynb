```python
class Nodo:
    """Representa un nodo con un dato y una referencia al siguiente."""
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None

class ListaEnlazada:
    """Gestiona la cabeza de la lista y las operaciones básicas."""
    def __init__(self):
        self.cabeza = None
        
    def agregar_inicio(self, dato):
        """Agrega un nodo al inicio (útil para construir la lista)."""
        nuevo_nodo = Nodo(dato)
        nuevo_nodo.siguiente = self.cabeza
        self.cabeza = nuevo_nodo
        
    def __str__(self):
        """Representación en cadena para imprimir la lista."""
        elementos = []
        actual = self.cabeza
        while actual:
            elementos.append(str(actual.dato))
            actual = actual.siguiente
        return " -> ".join(elementos)

def lista_inversa(lista_original):
    """
    Recorre la lista enlazada original y construye una nueva lista en orden inverso.

    Args:
        lista_original (ListaEnlazada): La lista enlazada a invertir.

    Returns:
        ListaEnlazada: Una nueva lista enlazada con los elementos en orden inverso.
    """
    
    # 1. Crear una nueva lista enlazada vacía
    lista_inversa = ListaEnlazada()
    
    # 2. Empezar a recorrer la lista original desde la cabeza
    actual = lista_original.cabeza
    
    # 3. Recorrer todos los nodos
    while actual:
        # La clave: Para invertir el orden, se inserta cada elemento *al inicio* de la nueva lista.
        # El último elemento de la original se convierte en la nueva cabeza.
        lista_inversa.agregar_inicio(actual.dato)
        
        # Mover al siguiente nodo
        actual = actual.siguiente
        
    return lista_inversa

# -----------------------------------------------------------
## Ejemplo de Uso
# -----------------------------------------------------------

# 1. Crear la lista enlazada original: 1 -> 2 -> 3
lista_original = ListaEnlazada()
lista_original.agregar_inicio(3) # Lista: 3
lista_original.agregar_inicio(2) # Lista: 2 -> 3
lista_original.agregar_inicio(1) # Lista: 1 -> 2 -> 3

print(f"Lista Original: {lista_original}")

# 2. Llamar a la función para obtener la lista inversa
lista_resultante = lista_inversa(lista_original)

# 3. Imprimir el resultado
print(f"Lista Inversa:  {lista_resultante}")
```
