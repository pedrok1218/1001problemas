```python
def suma_recursiva_lista(lista):
    """
    Calcula la suma de los valores en una lista usando recursión.

    Args:
        lista: La lista de números a sumar.

    Returns:
        La suma total de los valores en la lista.
    """
    # Caso Base: Si la lista está vacía, la suma es 0.
    if not lista:
        return 0
    # Caso Recursivo: Suma el primer elemento (lista[0]) con la suma del resto de la lista (lista[1:]).
    else:
        return lista[0] + suma_recursiva_lista(lista[1:])

# --- Ejemplo de Uso 🧑‍💻 ---

numeros = [10, 5, 8, 2]
resultado = suma_recursiva_lista(numeros)

print(f"La lista a sumar es: {numeros}")
print(f"La suma recursiva es: {resultado}") # Output: 25
```

La función sumarecursivalista sigue dos principios fundamentales de la recursión:
Caso Base: La condición de parada es cuando la lista está vacía if not lista. En este punto, la función devuelve 0, que es la suma de una lista sin elementos.
Paso Recursivo: El problema se reduce en cada llamada. La función toma el primer elemento {lista}[0] y lo suma al resultado de la misma función aplicada al resto de la lista {lista}[1:]. Esto asegura que la lista se haga progresivamente más pequeña hasta alcanzar el caso base.
