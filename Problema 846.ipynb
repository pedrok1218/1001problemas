```python
def posicion_menor_recursiva(lista, indice_actual=0, mejor_indice=0):
    """
    Encuentra la posición (índice) del menor valor en una lista de forma recursiva.

    Args:
        lista (list): La lista de números a buscar.
        indice_actual (int): El índice del elemento que se está examinando actualmente.
        mejor_indice (int): El índice del menor valor encontrado hasta ahora.

    Returns:
        int: La posición (índice) del menor valor en la lista.
    """
    # 1. Caso Base: Hemos llegado al final de la lista
    if indice_actual == len(lista):
        return mejor_indice

    # 2. Paso Condicional: Comprobar si el valor actual es menor que el mejor valor
    #    encontrado hasta ahora.
    if lista[indice_actual] < lista[mejor_indice]:
        mejor_indice = indice_actual  # Actualizar el mejor índice

    # 3. Paso Recursivo: Llamar a la función para el siguiente índice
    return posicion_menor_recursiva(lista, indice_actual + 1, mejor_indice)

# --- Ejemplo de Uso 🧑‍💻 ---

numeros_a = [5, 3, 9, 1, 6, 2]
posicion_a = posicion_menor_recursiva(numeros_a)

print(f"Lista: {numeros_a}")
print(f"El menor valor es {numeros_a[posicion_a]} y su posición es: {posicion_a}") # Output: 3 (El valor 1)

# Otro ejemplo
numeros_b = [40, 15, 20, 15]
posicion_b = posicion_menor_recursiva(numeros_b)

# Nota: En caso de empate (como el 15), devuelve la posición de la primera aparición.
print(f"Lista: {numeros_b}")
print(f"El menor valor es {numeros_b[posicion_b]} y su posición es: {posicion_b}") # Output: 1 (El primer valor 15)
```

La recursión funciona de la siguiente manera:

Caso Base: La función se detiene cuando el indice_actual (el que estamos examinando) alcanza el tamaño total de la lista (len(lista)). En ese momento, devuelve el mejor_indice que encontró durante todas las llamadas previas.

Parámetros Auxiliares: Necesitamos pasar los índices a través de las llamadas recursivas para mantener el estado de la búsqueda:

indice_actual: Se incrementa en 1 en cada llamada. Recorre la lista de inicio a fin.

mejor_indice: Guarda el índice donde se encuentra el valor más pequeño hasta ese momento de la recursión.

Comparación: En cada paso, se compara el valor en la posición indice_actual con el valor en la posición mejor_indice. Si el valor actual es menor, mejor_indice se actualiza.

Llamada Recursiva: La función se llama a sí misma para procesar el siguiente elemento (indice_actual + 1) y pasa el mejor_indice (actualizado o no) para la siguiente etapa.
