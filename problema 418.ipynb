```python
def combinar_y_ordenar_listas(lista1, lista2):
    """
    Combina dos listas de enteros y devuelve una NUEVA lista con todos 
    los elementos ordenados en orden ascendente.

    Args:
        lista1 (list): La primera lista de enteros.
        lista2 (list): La segunda lista de enteros.

    Returns:
        list: Una nueva lista con los elementos combinados y ordenados.
    """
    
    # 1. Concatenar las dos listas
    lista_combinada = lista1 + lista2
    
    # 2. Ordenar la lista combinada
    # sorted() devuelve una nueva lista ordenada (ascendente por defecto).
    return sorted(lista_combinada)

# --- Ejemplo de Uso ---

# Las listas de entrada no necesitan estar ordenadas
lista_a = [5, 1, 9, 3]
lista_b = [10, 4, 7, 2]

lista_final = combinar_y_ordenar_listas(lista_a, lista_b)

print(f"Lista 1: {lista_a}")
print(f"Lista 2: {lista_b}")
print(f"Lista Combinada y Ordenada: {lista_final}")

# Output: [1, 2, 3, 4, 5, 7, 9, 10]
```

El código utiliza dos operaciones: la concatenación de listas y la función sorted().

Operación: Se utiliza el operador de suma (+) entre las dos listas.

Resultado: Crea una nueva lista llamada lista_combinada que contiene todos los elementos de lista1 seguidos de todos los elementos de lista2.

Ejemplo: Si lista1 = [5, 1] y lista2 = [10, 4], lista_combinada será [5, 1, 10, 4].

Función sorted(): Esta es una función diseñada para ordenar colecciones.
Toma la lista desordenada (lista_combinada) como entrada.

Devuelve una nueva lista con todos sus elementos ordenados.

Por defecto, la ordenación es ascendente (de menor a mayor), que es lo que el programa requiere.
