```python
def contar_palabras_simple(nombre_archivo):
    """
    Abre un archivo, cuenta el número de palabras usando split() y len(),
    y muestra el total. Maneja el error si el archivo no existe.
    """
    try:
        # 1. Abrir el archivo y leer todo su contenido.
        with open(nombre_archivo, 'r') as archivo:
            contenido = archivo.read()
            
            # 2. Usar split() para dividir el contenido en una lista de "palabras".
            palabras = contenido.split()
            
            # 3. Usar len() para obtener el conteo de palabras.
            numero_palabras = len(palabras)
            
            # 4. Mostrar el total.
            print(f"✅ El archivo '{nombre_archivo}' contiene {numero_palabras} palabras.")
            
    except FileNotFoundError:
        print(f"❌ Error: El archivo '{nombre_archivo}' no fue encontrado.")

```
