```python
class Nodo:
    """Representa un nodo con un dato y una referencia al siguiente."""
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None

class ListaEnlazada:
    """Gestiona la cabeza de la lista y las operaciones básicas."""
    def __init__(self):
        self.cabeza = None
        
    def agregar_inicio(self, dato):
        """Método simple para construir la lista."""
        nuevo_nodo = Nodo(dato)
        nuevo_nodo.siguiente = self.cabeza
        self.cabeza = nuevo_nodo
        
    def contar_elementos(self):
        """Calcula el número total de elementos (n) en la lista."""
        contador = 0
        actual = self.cabeza
        while actual:
            contador += 1
            actual = actual.siguiente
        return contador

def subconjuntos_impares(lista_enlazada):
    """
    Calcula el número de subconjuntos de la lista con un número impar de elementos.

    El número total de subconjuntos de un conjunto de 'n' elementos es 2^n. 
    Exactamente la mitad de estos tendrán un número impar de elementos, es decir, 2^(n-1).
    """
    
    # 1. Contar el número total de elementos (n) en la lista enlazada.
    n = lista_enlazada.contar_elementos()
    
    # 2. Aplicar el principio combinatorio
    if n == 0:
        # Una lista vacía (n=0) sólo tiene el subconjunto vacío (0 elementos, par).
        return 0
    else:
        # Fórmula: 2^(n-1)
        # Usamos pow() para el cálculo de potencia.
        return pow(2, n - 1)
    
    # Crear y llenar la lista enlazada: [4, 3, 2, 1] (n=4)
mi_lista = ListaEnlazada()
mi_lista.agregar_inicio(4)
mi_lista.agregar_inicio(3)
mi_lista.agregar_inicio(2)
mi_lista.agregar_inicio(1)

# n = 4
n_elementos = mi_lista.contar_elementos()
resultado = subconjuntos_impares(mi_lista)

print(f"La lista tiene n = {n_elementos} elementos.")
print(f"El número de subconjuntos con tamaño impar es 2^({n_elementos}-1) = {resultado}") 
# Resultado: 2^(4-1) = 2^3 = 8
```
