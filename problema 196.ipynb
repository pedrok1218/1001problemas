```python
def calcular_promedio(lista_numeros):
    """
    Calcula el promedio (media aritmética) de los números en una lista.

    Args:
        lista_numeros (list): Una lista de números (enteros o flotantes).

    Returns:
        float: El promedio de la lista, o 0.0 si la lista está vacía.
    """
    # 1. Caso Base: Verificar si la lista está vacía para evitar división por cero.
    if not lista_numeros:
        return 0.0

    # 2. Calcular la suma de todos los elementos
    # La función sum() de Python es la forma más eficiente de hacer esto.
    suma_total = sum(lista_numeros)

    # 3. Obtener el número de elementos
    # La función len() devuelve la cantidad de elementos en la lista.
    cantidad_elementos = len(lista_numeros)

    # 4. Calcular el promedio: Suma total / Cantidad de elementos
    promedio = suma_total / cantidad_elementos

    return promedio

# --- Ejemplo de Uso ---

mis_notas = [8.5, 7.0, 9.5, 6.0, 10.0]
resultado_promedio = calcular_promedio(mis_notas)

print(f"La lista de números es: {mis_notas}")
print(f"La suma total es: {sum(mis_notas)}")
print(f"El número de elementos es: {len(mis_notas)}")
print(f"El promedio de la lista es: {resultado_promedio:.2f}") # Muestra con 2 decimales

# Ejemplo con una lista vacía
lista_vacia = []
promedio_vacio = calcular_promedio(lista_vacia)
print(f"\nEl promedio de una lista vacía es: {promedio_vacio}")
```

El programa aplica esto usando dos funciones integradas de Python para mayor eficiencia:

sum(lista): Obtiene rápidamente la suma de todos los elementos.

len(lista): Obtiene el número total de elementos.
