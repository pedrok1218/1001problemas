```python
def algoritmo_euclides(a, b):
    """
    Calcula el Máximo Común Divisor (MCD) de dos números enteros no negativos.
    
    Args:
        a (int): El primer número.
        b (int): El segundo número.
        
    Returns:
        int: El Máximo Común Divisor de a y b.
    """
    # Usamos la función abs() para manejar números negativos y asegurarnos de que el MCD sea positivo.
    a = abs(a)
    b = abs(b)
    
    # El algoritmo de Euclides mediante un bucle while
    while b:
        # En cada paso, 'a' se convierte en 'b' y 'b' se convierte en el resto de la división (a % b)
        a, b = b, a % b
    
    # Cuando 'b' se vuelve 0, 'a' contiene el MCD
    return a

def calcular_mcd_usuario():
    """
    Solicita dos números al usuario y muestra su MCD.
    """
    print("--- Calculadora del Máximo Común Divisor (MCD) ---")
    
    try:
        # 1. Leer los números del usuario
        num1 = int(input("Ingresa el primer número entero (a): "))
        num2 = int(input("Ingresa el segundo número entero (b): "))
        
        # 2. Manejar el caso especial: si ambos son cero, el MCD no está bien definido,
        #    pero por convención se puede devolver 0.
        if num1 == 0 and num2 == 0:
            print("El MCD de (0, 0) no está bien definido. Se devolverá 0.")
            mcd = 0
        else:
            # 3. Calcular el MCD
            mcd = algoritmo_euclides(num1, num2)
        
        # 4. Mostrar el resultado
        print(f"\nEl MCD de {num1} y {num2} es: {mcd}")
        
    except ValueError:
        print("❌ Error: Por favor, ingresa solo números enteros válidos.")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")

# Iniciar el programa
calcular_mcd_usuario()
```

Inicialización: Si el usuario ingresa (12, 8), entonces a=12 y b=8
Iteración 1:
El bucle se ejecuta porque $b \neq 0$.

Se calcula el resto: $12 \text{ mod } 8 = 4$.

La línea a, b = b, a % b actualiza las variables: $a$ se convierte en $8$ y $b$ se convierte en $4$.

Iteración 2:

El bucle se ejecuta porque $b=4$.

Se calcula el resto: $8 \text{ mod } 4 = 0$.

Actualización: $a$ se convierte en $4$ y $b$ se convierte en $0$.

Fin del Bucle: El bucle termina porque $b$ es $0$.

Resultado: La función devuelve el valor actual de $a$, que es $4$.
