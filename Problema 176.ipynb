```python
def calcular_valor_absoluto():
    """
    Solicita un número al usuario y calcula su valor absoluto.
    """
    try:
        # 1. Solicitar el número al usuario
        entrada = input("Ingresa un número (positivo o negativo): ")
        numero = float(entrada) # Usamos float para manejar enteros y decimales

        # 2. Calcular el valor absoluto usando la función abs()
        valor_absoluto = abs(numero)

        # 3. Mostrar el resultado
        print("\n--- Resultado ---")
        print(f"El número ingresado fue: {numero}")
        print(f"Su valor absoluto es: {valor_absoluto}")

    except ValueError:
        print("❌ Error: Por favor, ingresa una entrada numérica válida.")
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")

# Ejecutar el programa
calcular_valor_absoluto()
```

Si el número es positivo o cero, devuelve el mismo número (ej: $|5| = 5$).
Si el número es negativo, devuelve su equivalente positivo (ej: $|-5| = 5$).
