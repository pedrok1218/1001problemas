```python
class Nodo:
    """Representa un único nodo en la lista enlazada."""
    def __init__(self, valor, siguiente=None):
        self.valor = valor
        self.siguiente = siguiente

def construir_lista_unica_recursiva(nodo_actual, vistos=None):
    """
    Recorre recursivamente una lista enlazada y construye una nueva lista
    de Python (list) que solo contiene elementos únicos.

    :param nodo_actual: El nodo desde el cual comenzar el procesamiento.
    :param vistos: Un conjunto (set) para llevar un registro de los valores ya encontrados.
    :return: Una nueva lista de Python (list) con los valores únicos.
    """
    # Inicializar el conjunto 'vistos' en la primera llamada
    if vistos is None:
        vistos = set()

    # 1. Caso Base: Si el nodo actual es None (fin de la lista enlazada)
    if nodo_actual is None:
        return []

    # 2. Paso Recursivo
    
    # Llamada recursiva para procesar el resto de la lista enlazada
    # La cola_lista contiene los resultados únicos del resto de la lista.
    cola_lista = construir_lista_unica_recursiva(nodo_actual.siguiente, vistos)
    
    # 3. Procesar el valor del nodo actual
    
    if nodo_actual.valor not in vistos:
        # El valor es único:
        # a) Añadirlo al conjunto 'vistos'
        vistos.add(nodo_actual.valor)
        # b) Pre-cargar el valor al inicio de la lista devuelta por la recursión
        return [nodo_actual.valor] + cola_lista
    else:
        # El valor es duplicado:
        # Simplemente devolver la lista resultante de la recursión (omitiendo el valor actual)
        return cola_lista

# ----------------------------------------------------------------------
## Ejemplo de Uso

# Crear la lista enlazada: [1] -> [2] -> [3] -> [2] -> [4] -> [1] -> None
cabeza = Nodo(1)
cabeza.siguiente = Nodo(2)
cabeza.siguiente.siguiente = Nodo(3)
cabeza.siguiente.siguiente.siguiente = Nodo(2)  # Duplicado
cabeza.siguiente.siguiente.siguiente.siguiente = Nodo(4)
cabeza.siguiente.siguiente.siguiente.siguiente.siguiente = Nodo(1) # Duplicado

print("Lista enlazada original (valores): 1, 2, 3, 2, 4, 1")

# Ejecutar la función recursiva
lista_unica = construir_lista_unica_recursiva(cabeza)

print(f"Lista de Python con elementos únicos: {lista_unica}")
# El resultado será: [1, 2, 3, 4] (El orden de aparición en la lista enlazada se mantiene)
```
